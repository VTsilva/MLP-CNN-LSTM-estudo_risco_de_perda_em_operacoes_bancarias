O objetivo do Trabalho 1 é familiarizar o aluno com os processos de uso de diversos modelos baseados em redes neurais (MLP, CNN, LSTM) em contextos de previsão de séries temporais. Para isso, o estudante terá que procurar e escolher uma série temporal real que possua 2 ou mais registros (variáveis) ao longo de pelo menos 1000 pontos consecutivos.

Depois de filtrar e preparar esses dados, o estudante fará uma análise básica da série, plotando gráficos, determinando ciclos e tendências básicas. Em seguida, dividirá os dados em 70% treinamento, 20% validação, e 10% teste, e determinará o tamanho da janela móvel.

Ao contrário do que foi feito em aula, o estudante irá treinar os modelos acompanhando as funções de custo e acurácia dos conjuntos de treinamento e de validação, e irá determinar manualmente, por inspeção do gráfico, o número de épocas ideal, isto é, o instante em que o modelo começa a fazer o overfitting dos dados de treinamento, conforme aprendido no curso IAD-012. Fará então o retreinamento do modelo até o número de épocas ideal, e irá fazer o cálculo da acurácia nos conjuntos de treinamento, validação e teste. Se o número de épocas estimado para treinamento for julgado insuficiente, o estudante deverá continuar o treinamento até atingir o overfitting ou um plateau no valor da função de custo quando aplicado ao conjunto de validação.

Este processo será feito inicialmente só com uma das variáveis do dataset, a mesma sendo prevista. Deverão ser explorados todos os modelos indicados, concentrando a previsão somente no valor imediatamente a seguir da série de entrada (passo único). Ao final, deverá preparar uma comparação entre os modelos e identificar qual o melhor modelo que foi encontrado. Nesta comparação deverão ser usados elementos visuais que facilitem o entendimento das capacidades e deficiências dos diversos modelos.
